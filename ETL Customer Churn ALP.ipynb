# CREATE DIMENSION TABLE


```python
# Create Dim_Customer with corrected column names
dim_customer = df[['CustomerID', 'Gender', 'Senior Citizen', 'Partner', 'Dependents']].drop_duplicates()

# Save Dim_Customer to a CSV file
dim_customer.to_csv('dim_customer.csv', index=False)

# Display Dim_Customer
dim_customer.head()
```

```python
# Add unique LocationID for customers
df['LocationID'] = df['Zip Code']

# Customer Demographics Dimension
dim_location = df[['LocationID', 'Country', 'State', 'City']].drop_duplicates()

# Save to CSV
dim_location.to_csv('dim_location.csv', index=False)

# Display only the contents of the dim_location dimension
dim_location.head()
```

```python
# Add unique LocationID for customers
df['LocationID'] = df['Zip Code']

# Payment Method and Billing Preferences Dimension
dim_payment_method = df[['LocationID', 'Payment Method', 'Paperless Billing']].drop_duplicates()

# Save to CSV
dim_payment_method.to_csv('dim_payment_billing_dim.csv', index=False)

# Display only the contents of the dim_payment_method dimension
dim_payment_method.head()
```

```python
# Mapping unique combinations of Contract to numeric IDs
contract_mapping = {
    'Month-to-month': 10,
    'One year': 14,
    'Two year': 18
}

# Combining the Contract column to create combinations
df['Contract_Payment'] = df['Contract']

# Mapping the combinations to ContractID using the predefined mapping
df['ContractID'] = df['Contract_Payment'].map(contract_mapping)

# Saving only the Contract and ContractID columns to a CSV file named contract.csv
df[['Contract', 'ContractID']].to_csv('dim_contract.csv', index=False)

# Displaying only the Contract and ContractID columns
df[['Contract', 'ContractID']].head()
```

```python
import pandas as pd

# Define the mapping for services to numeric values
service_mapping = {
    'Yes': 1, 'No': 0,
    'DSL': 'D', 'Fiber optic': 'F', 'Cable': 'C',
    'None': 0  # For cases where the service is not available
}

# Function to convert a service value to its numeric equivalent or a default value
def map_service(value):
    return service_mapping.get(value, 2)  # Default to '2' for unknown or missing services

# Apply the service mapping for the relevant columns and create the ServiceID
service_columns = [
    'Phone Service', 'Multiple Lines', 'Internet Service', 'Online Security',
    'Online Backup', 'Device Protection', 'Tech Support'
]

df['ServiceID'] = df[service_columns].fillna('None').applymap(map_service).astype(str).agg(''.join, axis=1)

# Create the service dimension table by selecting distinct rows and writing to CSV
dim_service = df[['ServiceID'] + service_columns].drop_duplicates()

# Save the resulting DataFrame to a CSV file
dim_service.to_csv('dim_service.csv', index=False)

# Display the first few rows of the resulting 'dim_service' DataFrame
dim_service.head()
```

```python
# Combining the services owned by customers into one column
df['ServicePackage'] = df['Phone Service'] + '_' + df['Internet Service'] + '_' + \
                       df['Online Security'] + '_' + df['Online Backup'] + '_' + \
                       df['Device Protection'] + '_' + df['Tech Support'] + '_' + \
                       df['Streaming TV'] + '_' + df['Streaming Movies']

# Creating a unique ID for each service package
df['ServicePackageID'] = df['ServicePackage'].astype('category').cat.codes

# Creating the Service Package dimension that includes ServicePackageID and the service combinations
dim_service_package = df[['ServicePackageID', 'ServicePackage', 'Streaming TV', 'Streaming Movies']].drop_duplicates()

# Saving the Service Package dimension to CSV
dim_service_package.to_csv('dim_service_package.csv', index=False)

# Displaying the Service Package dimension
print(dim_service_package.head())
```
